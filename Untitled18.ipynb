```python
import os
import random
import torch
import torchaudio

SAMPLE_RATE = 16000
TEST_DURATION = 5

OUTPUT_DIR = "/content/drive/MyDrive/audio_separator_test_samples"

os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
def load_and_prepare(path, duration=5):

    waveform, sr = torchaudio.load(path)

    # Stereo -> mono
    if waveform.shape[0] > 1:
        waveform = waveform.mean(
            dim=0,
            keepdim=True
        )

    # Resample
    if sr != SAMPLE_RATE:

        resampler = torchaudio.transforms.Resample(
            sr,
            SAMPLE_RATE
        )

        waveform = resampler(waveform)

    target_length = SAMPLE_RATE * duration

    # Crop if longer
    if waveform.shape[1] > target_length:

        start = random.randint(
            0,
            waveform.shape[1] - target_length
        )

        waveform = waveform[
            :,
            start:start + target_length
        ]

    # Pad if shorter
    elif waveform.shape[1] < target_length:

        padding = target_length - waveform.shape[1]

        waveform = torch.nn.functional.pad(
            waveform,
            (0, padding)
        )

    return waveform
```

```python
def mix_at_snr(speech, noise, snr_db):

    speech_power = torch.mean(
        speech ** 2
    )

    noise_power = torch.mean(
        noise ** 2
    )

    desired_noise_power = (
        speech_power /
        (10 ** (snr_db / 10))
    )

    scale = torch.sqrt(
        desired_noise_power /
        (noise_power + 1e-8)
    )

    scaled_noise = noise * scale

    mixture = speech + scaled_noise

    # Prevent clipping
    peak = mixture.abs().max()

    if peak > 0.99:
        mixture = mixture / peak * 0.99

    return mixture
```

```python
speech_files = [
    "/content/drive/MyDrive/audio-separation/data/speech/1088-134315-0000.flac",
    "/content/drive/MyDrive/audio-separation/data/speech/1088-134315-0001.flac",
    "/content/drive/MyDrive/audio-separation/data/speech/1088-134315-0002.flac",
]

noise_files = [
    "/content/drive/MyDrive/audio-separation/data/noise/noise_0002.wav",
    "/content/drive/MyDrive/audio-separation/data/noise/noise_0004.wav",
    "/content/drive/MyDrive/audio-separation/data/noise/noise_0006.wav",
]
```

```python
snrs = [10, 5, 0]

for i in range(3):

    speech = load_and_prepare(
        speech_files[i],
        duration=TEST_DURATION
    )

    noise = load_and_prepare(
        noise_files[i],
        duration=TEST_DURATION
    )

    mixture = mix_at_snr(
        speech,
        noise,
        snrs[i]
    )

    output_path = os.path.join(
        OUTPUT_DIR,
        f"test_sample_{i+1}_snr_{snrs[i]}db.wav"
    )

    torchaudio.save(
        output_path,
        mixture,
        SAMPLE_RATE
    )

    print(
        f"Created: {output_path}"
    )
```

```python
from IPython.display import Audio, display

for i in range(1, 4):

    path = os.path.join(
        OUTPUT_DIR,
        f"test_sample_{i}_snr_{snrs[i-1]}db.wav"
    )

    print(f"Sample {i}")

    display(
        Audio(
            path,
            rate=SAMPLE_RATE
        )
    )
```
